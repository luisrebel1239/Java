### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
tiendas = [
    ('Tienda 1', tienda),
    ('Tienda 2', tienda2),
    ('Tienda 3', tienda3),
    ('Tienda 4', tienda4)
]

print("suma de ventas:\n")

nombres_tiendas = []
totales_ventas = []

for nombre_tienda, dataframe_tienda in tiendas:
    total_ventas = dataframe_tienda['Precio'].sum()
    nombres_tiendas.append(nombre_tienda)
    totales_ventas.append(total_ventas)
    print(f"{nombre_tienda} Total de ventas: ${total_ventas:.2f}")
```

```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots()
ax.scatter(x = nombres_tiendas, y = totales_ventas, marker = "*", c = "green",
            edgecolors = "red", linewidths = .5, s=200)
plt.xlabel("Tiendas")
plt.ylabel("Total de Ventas")
plt.title("Total de Ventas por Tienda")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.bar(nombres_tiendas, totales_ventas, color='skyblue')
plt.xlabel("Tiendas")
plt.ylabel("Total de Ventas")
plt.title("Total de Ventas por Tienda (Gráfica de Barras)")
plt.xticks(rotation=65, ha='right')
plt.tight_layout()
plt.grid(axis='y', linestyle='--')
plt.show()

```

```python
plt.figure(figsize=(10, 6))
plt.plot(nombres_tiendas, totales_ventas, marker='o', linestyle='-', color='red')
plt.xlabel("Tiendas")
plt.ylabel("Total de Ventas")
plt.title("Total de Ventas por Tienda (Gráfica de Líneas)")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.grid(axis='y', linestyle='--')
plt.show()
```

# 2. Ventas por categoría

```python
tienda.columns
```

```python

print("Ventas por categoría en cada tienda:\n")

categoria_por_tienda = {}
all_categorias = set()

for nombre, df in tiendas:
    conteo = df['Categoría del Producto'].value_counts()
    categoria_por_tienda[nombre] = conteo
    all_categorias.update(conteo.index)
    print(f"{nombre}:")
    for categoria, cantidad in conteo.items():
        print(f"  {categoria}: {cantidad} ventas")
    print()

all_categorias = sorted(list(all_categorias))
nombres_tiendas = [nombre for nombre, _ in tiendas]
```

```python
plt.figure(figsize=(12, 7))
bar_width = 0.2
x = range(len(nombres_tiendas))

for i, categoria in enumerate(all_categorias):
    valores = [categoria_por_tienda[nombre].get(categoria, 0) for nombre in nombres_tiendas]
    plt.bar([pos + i * bar_width for pos in x], valores, bar_width, label=categoria)
```

```python
plt.figure(figsize=(12, 7))

for i, categoria in enumerate(all_categorias):
    valores = [categoria_por_tienda[nombre].get(categoria, 0) for nombre in nombres_tiendas]
    plt.scatter(nombres_tiendas, valores, label=categoria)

plt.title("Dispersión de Ventas por Categoría en Cada Tienda")
plt.xlabel("Tiendas")
plt.ylabel("Cantidad de Ventas")
plt.legend(title="Categoría", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.grid(True)
plt.show()

```

```python
plt.figure(figsize=(12, 7))

for categoria in all_categorias:
    valores = [categoria_por_tienda[nombre].get(categoria, 0) for nombre in nombres_tiendas]
    plt.plot(nombres_tiendas, valores, marker='o', label=categoria)

plt.title("Ventas por Categoría en Cada Tienda (Gráfico Lineal)")
plt.xlabel("Tiendas")
plt.ylabel("Cantidad de Ventas")
plt.legend(title="Categoría", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.grid(True)
plt.show()
```

# 3. Calificación promedio de la tienda


```python
nombres_tiendas = []
promedios_calificacion = []

print("Promedio de calificación por tienda:\n")

for nombre, df in tiendas:
    promedio = df['Calificación'].mean()
    nombres_tiendas.append(nombre)
    promedios_calificacion.append(promedio)
    print(f"{nombre}: {promedio:.2f}")




```

```python
nombres = nombres_tiendas
valores = promedios_calificacion

plt.figure(figsize=(8, 5))
bars = plt.bar(nombres, valores, color='lightcoral')
plt.axhline(y=5, color='gray', linestyle='--', label='Máximo (5)')
plt.ylim(0, 5.5)
plt.title('Calificación Promedio por Tienda')
plt.xlabel('Tiendas')
plt.ylabel('Calificación Promedio')
plt.legend()


for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.1, f"{height:.2f}",
             ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()




```

```python
plt.figure(figsize=(8, 5))
plt.plot(nombres_tiendas, promedios_calificacion, marker='o', linestyle='-', color='green')

plt.title("Promedio de Calificación por Tienda (Gráfico Lineal)")
plt.xlabel("Tiendas")
plt.ylabel("Calificación Promedio")
plt.ylim(0, 5)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(nombres_tiendas, promedios_calificacion, color='red', s=100)

plt.title("Promedio de Calificación por Tienda (Gráfico de Dispersión)")
plt.xlabel("Tiendas")
plt.ylabel("Calificación Promedio")
plt.ylim(0, 5)
plt.grid(True)
plt.tight_layout()
plt.show()
```

# 4. Productos más y menos vendidos

```python
print("Productos más y menos vendidos por tienda:\n")

for nombre, df in tiendas:
    conteo = df['Producto'].value_counts()

    producto_mas_vendido = conteo.idxmax()
    ventas_max = conteo.max()

    producto_menos_vendido = conteo.idxmin()
    ventas_min = conteo.min()

    print(f"{nombre}:")
    print(f"  Más vendido: {producto_mas_vendido} ({ventas_max} ventas)")
    print(f"  Menos vendido: {producto_menos_vendido} ({ventas_min} venta(s))\n")


```

```python
nombres_tiendas = []
productos_mas_vendidos = []
ventas_maximas = []

productos_menos_vendidos = []
ventas_minimas = []

for nombre, df in tiendas:
    conteo = df['Producto'].value_counts()

    producto_mas_vendido = conteo.idxmax()
    ventas_max = conteo.max()

    producto_menos_vendido = conteo.idxmin()
    ventas_min = conteo.min()

    nombres_tiendas.append(nombre)
    productos_mas_vendidos.append(producto_mas_vendido)
    ventas_maximas.append(ventas_max)

    productos_menos_vendidos.append(producto_menos_vendido)
    ventas_minimas.append(ventas_min)

```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(len(nombres_tiendas))
bar_width = 0.35

plt.figure(figsize=(10, 6))
bars_max = plt.bar(x - bar_width/2, ventas_maximas, bar_width, label='Más vendidos', color='green')
bars_min = plt.bar(x + bar_width/2, ventas_minimas, bar_width, label='Menos vendidos', color='red')

for i, bar in enumerate(bars_max):
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.2,
             productos_mas_vendidos[i], ha='center', va='bottom', fontsize=9, rotation=45)

for i, bar in enumerate(bars_min):
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.2,
             productos_menos_vendidos[i], ha='center', va='bottom', fontsize=9, rotation=45)

plt.xticks(x, nombres_tiendas)
plt.ylabel("Cantidad de Ventas")
plt.title("Productos Más y Menos Vendidos por Tienda")
plt.legend()
plt.grid(axis='y')
plt.tight_layout()
plt.show()


```

```python
plt.figure(figsize=(10, 6))

plt.plot(nombres_tiendas, ventas_maximas, marker='o', label='Más vendidos', color='green')

for i, nombre in enumerate(nombres_tiendas):
    plt.text(nombre, ventas_maximas[i] + 0.2, productos_mas_vendidos[i],
             ha='center', va='bottom', fontsize=9, rotation=45)

plt.plot(nombres_tiendas, ventas_minimas, marker='o', label='Menos vendidos', color='red')
for i, nombre in enumerate(nombres_tiendas):
    plt.text(nombre, ventas_minimas[i] + 0.2, productos_menos_vendidos[i],
             ha='center', va='bottom', fontsize=9, rotation=45)

plt.ylabel("Cantidad de Ventas")
plt.title("Productos Más y Menos Vendidos por Tienda (Gráfico Lineal)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
plt.figure(figsize=(10, 6))


plt.scatter(nombres_tiendas, ventas_maximas, color='green', s=100, label='Más vendidos')
for i, nombre in enumerate(nombres_tiendas):
    plt.text(nombre, ventas_maximas[i] + 0.2, productos_mas_vendidos[i],
             ha='center', va='bottom', fontsize=9, rotation=45)

plt.scatter(nombres_tiendas, ventas_minimas, color='red', s=100, label='Menos vendidos')
for i, nombre in enumerate(nombres_tiendas):
    plt.text(nombre, ventas_minimas[i] + 0.2, productos_menos_vendidos[i],
             ha='center', va='bottom', fontsize=9, rotation=45)

plt.ylabel("Cantidad de Ventas")
plt.title("Productos Más y Menos Vendidos por Tienda (Dispersión)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

# 5. Envío promedio por tienda

```python

print("Costo de envío promedio por tienda:\n")
promedios_envio = {}

for nombre, df in tiendas:
    promedio = df['Costo de envío'].mean()
    promedios_envio[nombre] = promedio
    print(f"{nombre}: ${promedio:.2f}")
```

```python
nombres_tiendas = []
costos_promedio_envio = []

for nombre, df in tiendas:
    promedio = df['Costo de envío'].mean()
    nombres_tiendas.append(nombre)
    costos_promedio_envio.append(promedio)

```

```python
plt.figure(figsize=(8, 5))
bars = plt.bar(nombres_tiendas, costos_promedio_envio, color='skyblue')

for bar, costo in zip(bars, costos_promedio_envio):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.2,
             f"${costo:.2f}", ha='center', va='bottom')

plt.title("Costo de Envío Promedio por Tienda")
plt.xlabel("Tiendas")
plt.ylabel("Costo Promedio ($)")
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(nombres_tiendas, costos_promedio_envio, marker='o', linestyle='-', color='orange')

for i, costo in enumerate(costos_promedio_envio):
    plt.text(nombres_tiendas[i], costo + 0.2, f"${costo:.2f}", ha='center')

plt.title("Costo de Envío Promedio por Tienda (Línea)")
plt.xlabel("Tiendas")
plt.ylabel("Costo Promedio ($)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(nombres_tiendas, costos_promedio_envio, color='purple', s=100)


for i, costo in enumerate(costos_promedio_envio):
    plt.text(nombres_tiendas[i], costo + 0.2, f"${costo:.2f}", ha='center')

plt.title("Costo de Envío Promedio por Tienda (Dispersión)")
plt.xlabel("Tiendas")
plt.ylabel("Costo Promedio ($)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
